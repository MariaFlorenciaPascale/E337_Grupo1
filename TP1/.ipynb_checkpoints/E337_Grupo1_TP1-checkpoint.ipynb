### Verificamos como están compuestas las variables de los dataframes

```python
# Pasamos todas las variables a mayúsculas para armonizarlas:
df_2005_raw.columns = df_2005_raw.columns.str.upper().str.strip()
df_2025_raw.columns = df_2025_raw.columns.str.upper().str.strip()

# Elegimos variables:
variables = ['ANO4','PONDERA','CH04','CH06','CH07','CH08','NIVEL_ED',
             'ESTADO','CAT_INAC','IPCF','CAT_OCUP','PP07H','P21','PP04A','PP04C']

# Tamaño de las variables
print("\n TAMAÑO DE CADA BASE (sin filtrar)")
print(f"  2005: {df_2005_raw.shape[0]:,} filas x {df_2005_raw.shape[1]} columnas")
print(f"  2025: {df_2025_raw.shape[0]:,} filas x {df_2025_raw.shape[1]} columnas")

# Tipos de datos dentro de las variables
print("\n TIPO DE DATO POR VARIABLE")
print(f"  {'Variable':<12}  {'dtype 2005':>12}  {'dtype 2025':>12}")
for v in variables:
    d05 = str(df_2005_raw[v].dtype) if v in df_2005_raw.columns else "AUSENTE"
    d25 = str(df_2025_raw[v].dtype) if v in df_2025_raw.columns else "AUSENTE"
    flag = "  <<< DISTINTOS" if d05 != d25 else ""
    print(f"  {v:<12}  {d05:>12}  {d25:>12}{flag}")

# Valores únicos por variable 
print("\n VALORES ÚNICOS POR VARIABLE")
for v in variables:
    u05 = sorted(df_2005_raw[v].dropna().unique().tolist()) if v in df_2005_raw.columns else []
    u25 = sorted(df_2025_raw[v].dropna().unique().tolist()) if v in df_2025_raw.columns else []
    print(f"\n  {v}")
    print(f"    2005: {u05[:20]}{'...' if len(u05) > 20 else ''}")
    print(f"    2025: {u25[:20]}{'...' if len(u25) > 20 else ''}")


# Para cada variable categórica, verificamos que los mismos números representen lo mismo en 2005 y 2025 según el diccionario EPH.

print("\n CONSISTENCIA DE CÓDIGOS ENTRE 2005 Y 2025")

# Este diccionario es para chequear que los labels y valores sean los mismos que en los pdfs de referencia:
codigos_esperados = {
    'REGION'  : {1: 'Gran Buenos Aires', 40: 'NOA', 41: 'NEA',
                 42: 'Cuyo', 43: 'Pampeana', 44: 'Patagónica'},
    'CH04'    : {1: 'Varón', 2: 'Mujer'},
    'CH07'    : {1: 'Unido', 2: 'Casado', 3: 'Separado/Divorciado',
                 4: 'Viudo', 5: 'Soltero'},
    'CH08'    : {1: 'Obra social', 2: 'Mutual/prepaga', 3: 'Seg. público',
                 4: 'Sin cobertura', 12: 'OS+mutual', 13: 'OS+seg.pub',
                 23: 'Mutual+seg.pub', 123: 'OS+mutual+seg.pub'},
    'NIVEL_ED': {1: 'Prim. incompleta', 2: 'Prim. completa',
                 3: 'Sec. incompleta',  4: 'Sec. completa',
                 5: 'Sup/Univ. incompleta', 6: 'Sup/Univ. completa',
                 7: 'Sin instrucción'},
    'ESTADO'  : {0: 'Entrev. no realizada', 1: 'Ocupado', 2: 'Desocupado',
                 3: 'Inactivo', 4: 'Menor de 10 años'},
    'CAT_INAC': {0: 'No aplica', 1: 'Jubilado/Pensionado', 2: 'Rentista',
                 3: 'Estudiante', 4: 'Ama de casa', 5: 'Menor de 6 años',
                 6: 'Discapacitado', 7: 'Otros'},
    'CAT_OCUP': {0: 'No aplica', 1: 'Patrón', 2: 'Cuenta propia',
                 3: 'Asalariado', 4: 'Trab. familiar sin remun.'},
    'PP07H'   : {0: 'No aplica', 1: 'Sí tiene descuento', 2: 'No tiene descuento'},
    'PP04A'   : {1: 'Estatal', 2: 'Privado', 3: 'Otro'},
    'PP04C'   : {1: '1 persona', 2: '2 personas', 3: '3 personas',
                 4: '4 personas', 5: '5 personas', 6: '6-10 personas',
                 7: '11-25 personas', 8: '26-40 personas', 9: '41-100 personas',
                 10: '101-200 personas', 11: '201-500 personas', 12: 'Más de 500'},
}


for v, mapping in codigos_esperados.items():
    col = v if v in df_2005_raw.columns else None
    if col is None:
        continue
    vals_05 = set(int(x) for x in df_2005_raw[col].dropna().unique()
                  if x not in [0, 9, 99, 999] or v in ['ESTADO','CAT_INAC','CAT_OCUP','PP07H','PP04C'])
    vals_25 = set(int(x) for x in df_2025_raw[col].dropna().unique()
                  if x not in [9, 99, 999])
    esperados = set(mapping.keys())

    extra_05 = vals_05 - esperados
    extra_25 = vals_25 - esperados

    status = "OK" if not extra_05 and not extra_25 else "REVISAR"
    print(f"\n  {v} [{status}]")
    for k, etiqueta in mapping.items():
        en_05 = "SI" if k in vals_05 else "no"
        en_25 = "SI" if k in vals_25 else "no"
        print(f"    {k:>4} = {etiqueta:<30}  2005:{en_05}  2025:{en_25}")
    if extra_05:
        print(f"    *** Valores extra en 2005 NO esperados: {extra_05}")
    if extra_25:
        print(f"    *** Valores extra en 2025 NO esperados: {extra_25}")


# Acá vemos cuántas N/As hay en las variables antes de hacer el merge. La idea es que después de hacer el merge nos volvamos a 
# fijar cuantas N/As hay, y si concuerdan, se hizo bien el merge.
print("\n % NaN POR VARIABLE (antes de limpiar)")
print(f"  {'Variable':<12}  {'NaN% 2005':>10}  {'NaN% 2025':>10}")
for v in variables:
    p05 = df_2005_raw[v].isnull().sum() / len(df_2005_raw) * 100
    p25 = df_2025_raw[v].isnull().sum() / len(df_2025_raw) * 100
    print(f"  {v:<12}  {p05:>9.1f}%  {p25:>9.1f}%")
```
